#📌 Extracción

```python
import requests
import	pandas as pd
```

```python
url = "https://raw.githubusercontent.com/ingridcristh/challenge2-data-science-LATAM/refs/heads/main/TelecomX_Data.json"
response = requests.get(url)
data = response.json()

df = pd.DataFrame(data)
df.head()
```

#🔧 Transformación

```python
df = pd.read_json(url)
df.head()
```

```python
df = pd.json_normalize(data)
df
```

```python
df.info()
```

##Verificaciones

###Verificación de valores únicos

```python
for col in df.columns:
    print(f'Valores únicos en la columna {col}:{df[col].nunique()}')

    if df[col].nunique() < 50:
      print(f'{col}:{df[col].unique()}')
      print('-' * 50)
```

###Valores duplicados

```python
print("Número de duplicados", df.duplicated().sum())
```

###Valores nulos

```python
print("Número de nulos\n", df.isnull().sum())
```

###Valores vacíos o en blanco

```python
df.apply(lambda x: x.astype(str).str.strip() == '').sum()
```

###Cambiando la columma account.Charges.Total para float

```python
df['account.Charges.Total'] = df['account.Charges.Total'].apply(pd.to_numeric,errors='coerce')
print(df['account.Charges.Total'].dtype)
```

###Eliminación de valores vacíos em la columna Churn

```python
df = df[df['Churn'].str.strip() != '']
print(f'Números de filas después de eliminar los vacíos en Churn', len(df))
```

```python
df = df.dropna(subset=['account.Charges.Total'])
```

```python
print("Número de nulos\n", df.isnull().sum())
```

###**Opcional Creando cuentas diarias**

```python
df['cuentas_diarias'] = df['account.Charges.Monthly'] / 30
df
```

```python
columnas = {                                                        # Diccionario con los nuevos nombres de las columnas.
    'customerID': 'id',
    'customer.gender': 'genero',
    'customer.SeniorCitizen': 'tiene +60',
    'customer.Partner': 'posee_pareja',
    'customer.Dependents': 'posee_dependientes',
    'customer.tenure': 'tiempo_contrato',
    'phone.PhoneService': 'servicio_telefono',
    'phone.MultipleLines': 'multiples_lineas',
    'internet.InternetService': 'tipo_internet',
    'internet.OnlineSecurity': 'seguridad_online',
    'internet.OnlineBackup': 'backup_online',
    'internet.DeviceProtection': 'proteccion_dispositivo',
    'internet.TechSupport': 'soporte_tecnico',
    'internet.StreamingTV': 'streaming_tv',
    'internet.StreamingMovies': 'streaming_peliculas',
    'account.Contract': 'tipo_contrato',
    'account.PaperlessBilling': 'factura_digital',
    'account.PaymentMethod': 'metodo_pago',
    'account.Charges.Monthly': 'valor_mensal',
    'account.Charges.Total': 'total_cobrado'
}
df = df.rename(columns= columnas)
df
```

```python
df['Churn'] = df['Churn'].replace({'No': 'No', 'Yes': 'Sí'})
df['genero'] = df['genero'].replace({'Female': 'Femenino', 'Male': 'Masculino'})
df['posee_pareja'] = df['posee_pareja'].replace({'Yes': 'Sí', 'No': 'No'})
df['posee_dependientes'] = df['posee_dependientes'].replace({'Yes': 'Sí', 'No': 'No'})
df['servicio_telefono'] = df['servicio_telefono'].replace({'Yes': 'Sí', 'No': 'No'})
df['multiples_lineas'] = df['multiples_lineas'].replace({'No': 'No', 'Yes': 'Sí', 'No phone service': 'Sin servicio de teléfono'})
df['tipo_internet'] = df['tipo_internet'].replace({'No': 'No'})
df['tipo_contrato'] = df['tipo_contrato'].replace({'One year': 'Anual', 'Month-to-month': 'Mensual', 'Two year': 'Bienal'})
df.head()
```

```python
columnas_a_traducir = ['seguridad_online', 'backup_online', 'proteccion_dispositivo', 'soporte_tecnico', 'streaming_tv', 'streaming_peliculas']
mapeo = {'No': 'No', 'Yes': 'Sí', 'No internet service': 'Sin servicio de internet'}

for col in columnas_a_traducir:
    df[col] = df[col].replace(mapeo)

df.head()
```

```python
df['metodo_pago'] = df['metodo_pago'].replace({
    'Mailed check': 'Cheque enviado por correo',
    'Electronic check': 'Cheque electrónico',
    'Credit card (automatic)': 'Tarjeta de crédito (automático)',
    'Bank transfer (automatic)': 'Transferencia bancaria (automática)'
})
df.head()
```

**Distribución de la Evasión**

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 12))
churn_counts = df['Churn'].value_counts()
churn_percentages = (churn_counts / len(df)) * 100

ax = churn_counts.plot(kind='bar')
plt.title('Distribuicion de Evasion de Clientes')
plt.xlabel('Cancelamiento')
plt.ylabel('Cantidad de Clientes')
plt.xticks(rotation=0)

for i, percentage in enumerate(churn_percentages):
    ax.text(i, churn_counts.iloc[i] + 50, f'{percentage:.1f}%', ha='center', va='bottom')

plt.show()
```

**Conteo de evasión por variables numéricas**

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Renombrar la columna (de 'valor_mensal' a 'valor mensual')
df = df.rename(columns={'valor_mensal': 'valor mensual'})

# 2. Configuración estética global
sns.set_theme(style="white", palette="muted")

# IMPORTANTE: Aquí actualizamos la lista para que use el nombre con espacio
numericas = ['tiempo_contrato', 'valor mensual', 'total_cobrado', 'cuentas_diarias']
colores = {'No': '#4A90E2', 'Sí': '#FF6B6B'}

# 3. Creación de los gráficos
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for i, col in enumerate(numericas):
    # Gráfico principal: Boxplot
    sns.boxplot(
        data=df, x='Churn', y=col, ax=axes[i],
        hue='Churn', palette=colores,
        width=0.5, linewidth=1.5, fliersize=2, legend=False
    )

    # Títulos y etiquetas limpios
    # col.replace("_", " ") quitará los guiones de las otras variables automáticamente
    axes[i].set_title(col.replace("_", " ").title(), fontsize=14, fontweight='bold', pad=15)
    axes[i].set_xlabel('')
    axes[i].set_ylabel('')

    # Quitar los bordes de la "caja" del gráfico
    sns.despine(ax=axes[i], left=True)

    # Agregar una cuadrícula muy tenue solo en el eje Y
    axes[i].grid(axis='y', color='gray', linestyle='-', alpha=0.1)

plt.tight_layout(pad=3.0)
plt.show()
```
